```python

import math

def hav(a, b):
    return np.sin((b - a)/2)**2

def haversine(lat1, lon1, lat2, lon2):
    R = 6371  # Rayon de la Terre en kilomètres
    phi1 = math.radians(lat1)
    phi2 = math.radians(lat2)

    lamb1 = math.radians(lon1)
    lamb2 = math.radians(lon2)

    a = hav(phi1, phi2) + math.cos(phi1) * math.cos(phi2) * hav(lamb1, lamb2)
    d = 2 * R * math.asin(math.sqrt(a))

    return d
```

```python
import geopy.distance as geo

cols = doc['GeoJSON'].dropna()

trees = cols.apply(lambda x: json.loads(x)['coordinates'][::-1])

pair = trees.sample(n=2)

print(pair)

d = geo.distance((pair.iloc[0][0], pair.iloc[0][1]), (pair.iloc[1][0], pair.iloc[1][1]))
d2 = haversine(pair.iloc[0][0], pair.iloc[0][1], pair.iloc[1][0], pair.iloc[1][1])
print(d)
print(d2)
```

```python
maison = (45.084067788908314, 5.7715330528002795)

distances = trees.apply(lambda x: geo.distance(maison, (x[0], x[1])).km)

top = distances.nsmallest(10)
toptrees = trees.loc[top.index]

result = pd.DataFrame({'km': top, 'coords': trees.loc[top.index]})

print(result)
```

```python
from itertools import combinations
distances = pd.DataFrame(
    [(i, j, geo.distance(a, b).km)
     for (i, a), (j, b) in combinations(trees.head(100).items(), 2)],
    columns=['tree_1', 'tree_2', 'km'],
).sort_values('km')

print(distances['km'].mean())

full = pd.concat([
    distances,
    distances.rename(columns={'tree_1': 'tree_2', 'tree_2': 'tree_1'}),
])

par_arbre = full.groupby('tree_1')['km']
```
